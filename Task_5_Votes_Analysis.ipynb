## Task 5: Votes Analysis

### Objective
The objective of this task is to identify restaurants with the highest and lowest number of votes and analyze the relationship between the number of votes and restaurant ratings.

### Tools Used
- Python
- Pandas
- Matplotlib

### Dataset
Zomato Restaurant Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset_url = "https://raw.githubusercontent.com/radhika-patel017/SAM-AI-Tech-Task/main/Zomato-Dataset.csv"

df = pd.read_csv(dataset_url)

print("Dataset loaded successfully!")
```

```python
df = df.dropna(subset=["Cuisines"]).copy()
df.reset_index(drop=True, inplace=True)

print("Cleaned dataset shape:", df.shape)
```

```python
df.shape
```

```python
df[["Restaurant Name", "City", "Cuisines", "Aggregate rating", "Votes"]].head()
```

```python
highest_votes = df.sort_values("Votes", ascending=False).head(10)

highest_votes[
    ["Restaurant Name", "City", "Votes", "Aggregate rating"]
]
```

```python
lowest_votes = df.sort_values("Votes", ascending=True).head(10)

lowest_votes[
    ["Restaurant Name", "City", "Votes", "Aggregate rating"]
]
```

```python
lowest_non_zero_votes = (
    df[df["Votes"] > 0]
    .sort_values("Votes", ascending=True)
    .head(10)
)

lowest_non_zero_votes[
    ["Restaurant Name", "City", "Votes", "Aggregate rating"]
]
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    highest_votes["Restaurant Name"],
    highest_votes["Votes"]
)

plt.title("Top 10 Restaurants by Number of Votes")
plt.xlabel("Number of Votes")
plt.ylabel("Restaurant Name")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

### Insight - Highest Votes

**Toit in Bangalore** has the highest number of votes in the dataset with **10,934 votes** and an aggregate rating of **4.8**.

This indicates that Toit is one of the most highly engaged-with restaurants in the dataset.

```python
lowest_votes = df.sort_values("Votes", ascending=True).head(10)

lowest_votes[
    ["Restaurant Name", "City", "Votes", "Aggregate rating"]
]
```

### Insight - Lowest Votes

The minimum vote count in the dataset is **0**, and several restaurants share this value.

Many of these restaurants also have an aggregate rating of **0.0**, indicating that they are unrated.

```python
correlation_data = df[
    (df["Votes"] > 0) &
    (df["Aggregate rating"] > 0)
]
```

```python
correlation = correlation_data["Votes"].corr(
    correlation_data["Aggregate rating"]
)

print("Correlation between Votes and Rating:", round(correlation, 2))
```

```python
plt.figure(figsize=(9, 6))

plt.scatter(
    correlation_data["Votes"],
    correlation_data["Aggregate rating"],
    alpha=0.4
)

plt.title("Relationship Between Votes and Restaurant Rating")
plt.xlabel("Number of Votes")
plt.ylabel("Aggregate Rating")

plt.tight_layout()
plt.show()
```

## Final Insights - Task 5

The restaurant with the highest number of votes is **Toit in Bangalore**, with **10,934 votes** and an aggregate rating of **4.8**.

The minimum vote count in the dataset is **0**, which is shared by several restaurants. Many of these restaurants are also unrated.

For a more meaningful relationship analysis, restaurants with zero votes and zero aggregate ratings were excluded.

The correlation coefficient between the number of votes and restaurant ratings is approximately **0.41**.

This indicates a **moderate positive relationship** between votes and ratings. In general, restaurants receiving more votes tend to have somewhat higher ratings.

However, the relationship is not strong enough to conclude that a higher number of votes always leads to a higher restaurant rating.

The scatter plot supports this finding by showing a general positive pattern with considerable variation.

# Task 5 Conclusion

The votes analysis identified the restaurants with the highest and lowest levels of customer engagement and examined the relationship between votes and ratings.

**Toit in Bangalore** received the highest number of votes, with **10,934 votes** and a rating of **4.8**.
